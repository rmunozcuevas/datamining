```python
df['Discount'] = df['Discount'].str.extract(r'(\d+)').astype(float).fillna(0)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Sub Category'] = le.fit_transform(df['Sub Category'])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled = scaler.fit_transform(df[['Price', 'Discount']])

df_scaled = pd.DataFrame(scaled, columns=['Price', 'Discount'])
df_scaled['Sub Category'] = df['Sub Category'].values
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)
df_scaled['Cluster'] = kmeans.fit_predict(df_scaled[['Price', 'Discount']])
```

```python
df_scaled.groupby('Cluster')[['Price', 'Discount']].mean()
```

```python
df_result = df.copy()
df_result['Cluster'] = df_scaled['Cluster']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming df_scaled has 'Price', 'Discount', and 'Cluster'
plt.figure(figsize=(10,7))
sns.scatterplot(
    data=df_scaled,
    x='Price',
    y='Discount',
    hue='Cluster',
    palette='tab10',   # nice distinct colors
    s=100              # size of points
)

plt.title("Clusters of Products by Price and Discount")
plt.xlabel("Price (scaled)")
plt.ylabel("Discount (scaled)")
plt.legend(title='Cluster')
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    df_scaled['Price'],
    df_scaled['Discount'],
    df_scaled['Sub Category'],  # numeric encoding
    c=df_scaled['Cluster'],
    cmap='tab10',
    s=50
)

ax.set_xlabel('Price')
ax.set_ylabel('Discount')
ax.set_zlabel('Sub Category')
plt.title("3D Cluster Visualization")
plt.show()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
components = pca.fit_transform(df_scaled[['Price', 'Discount', 'Sub Category']])
df_scaled['PC1'] = components[:,0]
df_scaled['PC2'] = components[:,1]

sns.scatterplot(data=df_scaled, x='PC1', y='PC2', hue='Cluster', palette='tab10')
plt.title("Clusters (PCA-reduced)")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(df_scaled[['Price', 'Discount']], df_scaled['Cluster'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define readable labels for each cluster
cluster_labels = {
    0: "Low Price, Low Discount",
    1: "Mid Price, High Discount",
    2: "High Price, No Discount",
    3: "Med-High Price, No Discount"
}

# Add the labels to your result dataframe
df_result['Cluster Label'] = df_result['Cluster'].map(cluster_labels)

# --- Plot in real-world units ---
plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=df_result,
    x='Price',
    y='Discount',
    hue='Cluster Label',
    palette='tab10',
    s=100,
    edgecolor='black'
)

# Plot real cluster centers
plt.scatter(
    centers_df['Price'],
    centers_df['Discount'],
    c='black',
    s=300,
    alpha=0.6,
    marker='X',
    label='Centers'
)

# --- Styling ---
plt.title("Product Clusters by Price and Discount", fontsize=15, fontweight='bold')
plt.xlabel("Price ($)", fontsize=12)
plt.ylabel("Discount (%)", fontsize=12)
plt.legend(title="Cluster", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
